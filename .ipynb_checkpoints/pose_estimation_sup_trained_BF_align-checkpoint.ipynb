#### Reading Orthomosaic Map

```python
#-------------------- PATH --------------------
ORTHO_PATH = "../shared/datasets/garrulus-arnsberg-forest/data_collection/2025-11-19_field-D-E/post-processing-results/2025-11-19_field-D_flight-12_rgb-w-gcp/rgb_wide_images-only_orthomosaic.tif"
DSM_PATH = "../shared/datasets/garrulus-arnsberg-forest/data_collection/2025-11-19_field-D-E/post-processing-results/2025-11-19_field-D_flight-10_rgb-w-gcp/rgb_wide_images-only_dtm.tif"
# UAV image resolution in pixels (width x height)
uav_image_res_w = 1936
uav_image_res_h = 1460

# UAV GSD in m/px
gsd_uav_x_m = 0.6 / 100.0
gsd_uav_y_m = 0.6 / 100.0

#-------------------- CHECK ORTHOMOSAIC FILE --------------------
if not os.path.exists(ORTHO_PATH):
    raise FileNotFoundError(f"Orthomosaic not found: {ORTHO_PATH}")

#-------------------- OPEN ORTHOMOSAIC AND READ DATA --------------------
with rasterio.open(ORTHO_PATH) as ds:
    # Read the entire orthomosaic image 
    image_map_raw = ds.read(out_dtype="uint8")
    # get affine transform to convert pixel to map coordinates (UTM)
    px2utm_transform = ds.transform
    # get the Coordinate Reference System (CRS) of the orthomosaic
    ortho_crs = ds.crs
    # get image spatial bounds
    bounds = ds.bounds

    print("Orthomosaic CRS:", ortho_crs)
    print("Orthomosaic size (W x H):", ds.width, "x", ds.height)

    # calculate GSD from transform matrix (pixel size in meters)
    GSD_x = abs(px2utm_transform.a)
    GSD_y = abs(px2utm_transform.e)
    print(f"Orthomosaic GSD (m/pixel): ({GSD_x:.6f}, {GSD_y:.6f})")

#-------------------- PREPROCESS IMAGE BANDS --------------------
# keep the RGB bounds
if image_map_raw.shape[0] > 3:
    image_map_raw = image_map_raw[:3, :, :]

# convert image array to height x width x bands format for OpenCV
image_map_rgb = np.transpose(image_map_raw, (1, 2, 0))
# convert from RGB to BGR 
image_map_bgr = cv2.cvtColor(image_map_rgb, cv2.COLOR_RGB2BGR)

# get orthomosaic image dimensions
h_map, w_map = image_map_bgr.shape[:2]

#-------------------- TRANSFORM UAV GPS COORDINATES TO ORTHOMOSAIC CRS --------------------
# create transformer from WGS84 to orthomosaic CRS
transformer = Transformer.from_crs("EPSG:4326", ortho_crs, always_xy=True)
```

#### Helper Functions

```python
# -------------------- BASIC UTILS --------------------
def to_gray(bgr):
    return cv2.cvtColor(bgr, cv2.COLOR_BGR2GRAY)


def resize_keep(img, max_dim=MAX_DIM):
    h, w = img.shape[:2]
    scale = 1.0
    if max(h, w) > max_dim:
        scale = max_dim / float(max(h, w))
        img = cv2.resize(img, (int(w * scale), int(h * scale)), interpolation=cv2.INTER_AREA)
    return img, scale


def get_euler(R):
    sy = np.sqrt(R[0, 0]**2 + R[1, 0]**2)
    if sy > 1e-6:
        return np.degrees([
            np.arctan2(R[1, 0], R[0, 0]),
            np.arctan2(-R[2, 0], sy),
            np.arctan2(R[2, 1], R[2, 2])
        ])
    else:
        return np.degrees([
            0,
            np.arctan2(-R[2, 0], sy),
            np.arctan2(-R[1, 2], R[1, 1])
        ])


# -------------------- IMU → ROTATION --------------------
def R_from_quaternion(w, x, y, z):
    n = np.sqrt(w*w + x*x + y*y + z*z)
    w, x, y, z = w/n, x/n, y/n, z/n

    return np.array([
        [1 - 2*(y*y + z*z),     2*(x*y - z*w),     2*(x*z + y*w)],
        [    2*(x*y + z*w), 1 - 2*(x*x + z*z),     2*(y*z - x*w)],
        [    2*(x*z - y*w),     2*(y*z + x*w), 1 - 2*(x*x + y*y)]
    ], dtype=np.float64)


# -------------------- FAST NMS (CLEAN VERSION) --------------------
def nms_fast(corners, H, W, dist_thresh):
    """
    corners: shape (3, N) → [x, y, score]
    """
    if corners.shape[1] == 0:
        return corners, []

    # Sort by confidence DESC
    inds = np.argsort(-corners[2])
    corners = corners[:, inds]

    grid = np.zeros((H, W), dtype=bool)
    keep = []

    for i in range(corners.shape[1]):
        x, y = int(corners[0, i]), int(corners[1, i])

        if grid[y, x]:
            continue

        keep.append(i)

        x_min = max(0, x - dist_thresh)
        x_max = min(W, x + dist_thresh + 1)
        y_min = max(0, y - dist_thresh)
        y_max = min(H, y + dist_thresh + 1)

        grid[y_min:y_max, x_min:x_max] = True

    return corners[:, keep], keep


# -------------------- SUPERPOINT FEATURE EXTRACTION --------------------
@torch.no_grad()
def extract_features(net, img, conf_thresh=0.0005, nms_dist=4, max_kpts=1000):

    H, W = img.shape

    # --- INPUT ---
    inp = torch.from_numpy(img / 255.0).float()[None, None].to(device)

    # --- FORWARD ---
    out = net(inp)

    # SUPPORT BOTH NETWORK TYPES (IMPORTANT)
    if isinstance(out, dict):
        semi, desc = out["semi"], out["desc"]
    else:
        semi, desc = out

    # --- HEATMAP ---
    heatmap = flattenDetection(semi).cpu().numpy().squeeze()

    ys, xs = np.where(heatmap >= conf_thresh)

    if len(xs) == 0:
        return None, None, None

    pts = np.stack([xs, ys, heatmap[ys, xs]], axis=1).T

    # --- NMS ---
    pts_nms, _ = nms_fast(pts, H, W, nms_dist)
    pts_nms = pts_nms.T

    if pts_nms.shape[0] == 0:
        return None, None, None

    # --- LIMIT KEYPOINTS ---
    if pts_nms.shape[0] > max_kpts:
        idx = np.argsort(pts_nms[:, 2])[-max_kpts:]
        pts_nms = pts_nms[idx]

    # --- CONVERT ---
    kpts = torch.from_numpy(pts_nms[:, :2]).float().to(device)
    scores = torch.from_numpy(pts_nms[:, 2]).float().to(device)

    # --- SAMPLE DESCRIPTORS ---
    grid = kpts.clone()
    grid[:, 0] = grid[:, 0] / (W - 1) * 2 - 1
    grid[:, 1] = grid[:, 1] / (H - 1) * 2 - 1
    grid = grid.view(1, 1, -1, 2)

    desc = torch.nn.functional.grid_sample(desc, grid, align_corners=True)
    desc = torch.nn.functional.normalize(desc.reshape(1, 256, -1), dim=1)

    return kpts.unsqueeze(0), scores.unsqueeze(0), desc




# ---------- RESIZE FUNCTION ----------
def resize_for_sp(img, max_size=MAX_SP_SIZE):
    h, w = img.shape[:2]
    scale = max_size / max(h, w) if max(h, w) > max_size else 1.0
    return cv2.resize(img, (int(w*scale), int(h*scale))), scale

# ---------- MATCH VISUALIZATION ----------
def plot_matches(img0, img1, pts0, pts1, title="Matches", max_draw=100):
    if img0.ndim == 2:
        img0 = cv2.cvtColor(img0, cv2.COLOR_GRAY2RGB)
    if img1.ndim == 2:
        img1 = cv2.cvtColor(img1, cv2.COLOR_GRAY2RGB)

    H0, W0 = img0.shape[:2]
    H1, W1 = img1.shape[:2]

    canvas = np.zeros((max(H0, H1), W0 + W1, 3), dtype=np.uint8)
    canvas[:H0, :W0] = img0
    canvas[:H1, W0:] = img1

    plt.figure(figsize=(12, 8))
    plt.imshow(canvas)
    plt.axis("off")
    plt.title(title)

    num = min(len(pts0), max_draw)
    colors = plt.cm.jet(np.linspace(0, 1, num))

    for i in range(num):
        x0, y0 = pts0[i]
        x1, y1 = pts1[i]

        plt.plot([x0, x1 + W0], [y0, y1], color=colors[i], linewidth=1)
        plt.scatter(x0, y0, color=colors[i], s=10)
        plt.scatter(x1 + W0, y1, color=colors[i], s=10)

    plt.show()

# ---------- VISUALIZE INLIER MATCHES ----------
def plot_inlier_matches(img0, img1, pts0, pts1, title="Inlier Matches", max_draw=100):
    """
    pts0, pts1: only inlier matched keypoints (after RANSAC/Homography)
    """
    if img0.ndim == 2:
        img0 = cv2.cvtColor(img0, cv2.COLOR_GRAY2RGB)
    if img1.ndim == 2:
        img1 = cv2.cvtColor(img1, cv2.COLOR_GRAY2RGB)

    H0, W0 = img0.shape[:2]
    H1, W1 = img1.shape[:2]

    canvas = np.zeros((max(H0,H1), W0+W1, 3), dtype=np.uint8)
    canvas[:H0, :W0] = img0
    canvas[:H1, W0:W0+W1] = img1

    plt.figure(figsize=(12,8))
    plt.imshow(canvas)
    plt.axis("off")
    plt.title(title)

    num_draw = min(len(pts0), max_draw)
    colors = plt.cm.jet(np.linspace(0,1,num_draw))

    for i in range(num_draw):
        x0, y0 = pts0[i]
        x1, y1 = pts1[i]
        plt.plot([x0, x1+W0], [y0, y1], color=colors[i], linewidth=1)
        plt.scatter(x0, y0, color=colors[i], s=10)
        plt.scatter(x1+W0, y1, color=colors[i], s=10)

    plt.show()

```

#### Camera Intrinsics

```python
K = np.array([
    [1806.84457,     0.0,        945.416679],
    [     0.0,    1806.43529,    738.031572],
    [     0.0,        0.0,          1.0     ]
], dtype=np.float64)

dist_coeffs = np.array([-0.20484306, 0.1278626, -0.00041228, 0.0005107, -0.06076351]).reshape(-1, 1)
```

#### Pose Estimation Pipeline

```python
# ---------- CONFIG ----------
gsd_uav_x_m = 0.6 / 100.0
gsd_uav_y_m = 0.6 / 100.0


# ---------- LOOP OVER IMAGES ----------
image_dir = "../Field_D_13_images/extracted_images2/"
meta_path = "../Field_D_13_images/extracted_images2/metadata_with_euler.csv"
matches_out_dir = "matches_out_pt_trained_BF"
os.makedirs(matches_out_dir, exist_ok=True)

meta_data = pd.read_csv(meta_path)
image_paths = sorted(glob.glob(os.path.join(image_dir, "frame_*.png")))

results = []

for image_path in image_paths:

    filename = os.path.basename(image_path)
    uav_imag_num = int(filename.split("_")[1].split(".")[0])

    if uav_imag_num < 214:
        continue

    print("\nProcessing:", uav_imag_num, filename)

    # ---------- LOAD METADATA ----------
    row = meta_data.iloc[uav_imag_num]

    uav_lat_gps = float(row["gps_latitude"])
    uav_lon_gps = float(row["gps_longitude"])
    uav_alt_gps = float(row["gps_altitude"])

    uav_roll = float(row["roll"])
    uav_pitch = float(row["pitch"])
    uav_yaw = float(row["yaw"])

    qx = float(row["imu_orientation_x"])
    qy = float(row["imu_orientation_y"])
    qz = float(row["imu_orientation_z"])
    qw = float(row["imu_orientation_w"])

    # ---------- LOAD UAV IMAGE ----------
    img_uav_bgr = cv2.imread(image_path)
    if img_uav_bgr is None:
        print("Image read failed")
        continue
    img_uav_bgr = img_uav_bgr.astype("uint8")

    # ---------- ROTATE UAV IMAGE ----------
    yaw_deg = np.rad2deg(uav_yaw)
    rotation_angle = -90.0 + yaw_deg

    h, w = img_uav_bgr.shape[:2]
    center = (w/2.0, h/2.0)
    M = cv2.getRotationMatrix2D(center, rotation_angle, 1.0)
    cos = abs(M[0,0])
    sin = abs(M[0,1])
    new_w = int(h*sin + w*cos)
    new_h = int(h*cos + w*sin)
    M[0,2] += (new_w - w)/2.0
    M[1,2] += (new_h - h)/2.0
    img_uav_rot_bgr = cv2.warpAffine(img_uav_bgr, M, (new_w,new_h), flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_CONSTANT)

    # ---------- MAP COORDINATES ----------
    uav_e_map, uav_n_map = transformer.transform(uav_lon_gps, uav_lat_gps)
    row_f, col_f = rowcol(px2utm_transform, uav_e_map, uav_n_map)
    r = int(round(row_f))
    c = int(round(col_f))

    # ---------- PATCH EXTRACTION ----------
    ortho_per_uav_x = gsd_uav_x_m / GSD_x
    ortho_per_uav_y = gsd_uav_y_m / GSD_y
    crop_w = int(uav_image_res_w * ortho_per_uav_x) // 2
    crop_h = int(uav_image_res_h * ortho_per_uav_y) // 2
    r1 = max(r - crop_h, 0)
    r2 = min(r + crop_h, h_map)
    c1 = max(c - crop_w, 0)
    c2 = min(c + crop_w, w_map)
    patch_bgr = image_map_bgr[r1:r2, c1:c2].copy()
    if patch_bgr.size == 0:
        print("Patch extraction failed")
        continue

    # ---------- RESIZE ----------
    img1_rs, s1 = resize_keep(img_uav_rot_bgr)
    img2_rs, s2 = resize_keep(patch_bgr)
    img1_gray = to_gray(img1_rs)
    img2_gray = to_gray(img2_rs)

    # ---------- SUPERPOINT + BF MATCHING ----------
    k0, s0, d0 = extract_features(sp, img1_gray)
    k1, s1_feat, d1 = extract_features(sp, img2_gray)
    if k0 is None or k1 is None:
        print("Too few matches")
        continue
    kpts0 = k0[0].cpu().numpy()
    kpts1 = k1[0].cpu().numpy()
    desc0 = d0[0].permute(1,0).cpu().numpy()
    desc1 = d1[0].permute(1,0).cpu().numpy()


    # ---------- KNN MATCHING ----------
    bf = cv2.BFMatcher(cv2.NORM_L2)
    knn_matches = bf.knnMatch(desc0, desc1, k=2)

    good_matches_list = []
    for m, n in knn_matches:
        if m.distance < 0.75 * n.distance:  # RATIO_TEST, adjust if needed
            good_matches_list.append(m)

    mkpts0 = np.array([kpts0[m.queryIdx] for m in good_matches_list])
    mkpts1 = np.array([kpts1[m.trainIdx] for m in good_matches_list])
    matches0 = good_matches_list
    raw_matches = len(kpts0)
    good_matches = len(mkpts0)
    print("SuperPoint KNN matches:", good_matches)

    if good_matches < 20:
        print("Too few matches")
        continue


    # ---------- SCALE BACK TO ORIGINAL ----------
    mkpts0 /= s1
    mkpts1 /= s2

    # ---------- HOMOGRAPHY FILTER ----------
    src = mkpts0.reshape(-1,1,2)
    dst = mkpts1.reshape(-1,1,2)
    H, mask = cv2.findHomography(src, dst, cv2.RANSAC, 5.0)
    if H is None:
        print("Homography failed")
        continue
    mask = mask.ravel().astype(bool)
    mkpts0 = mkpts0[mask]
    mkpts1 = mkpts1[mask]
    if len(mkpts0) >= 4:
        src_pts = mkpts0.reshape(-1,1,2)
        dst_pts = mkpts1.reshape(-1,1,2)
        H, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, 3.0)
        maskH = mask.ravel().astype(bool)
    else:
        maskH = None

    # ---------- REBUILD MATCHES FOR DRAWING ----------
    matches_vis = [cv2.DMatch(_queryIdx=i, _trainIdx=i, _distance=0) 
                for i in range(len(mkpts0))]  # one-to-one for filtered inliers

    # ---------- SAVE MATCHED IMAGE VISUALIZATION ----------
    kpts0_cv = [cv2.KeyPoint(float(p[0]), float(p[1]), 1) for p in mkpts0]
    kpts1_cv = [cv2.KeyPoint(float(p[0]), float(p[1]), 1) for p in mkpts1]

    img_vis = cv2.drawMatches(
        img_uav_rot_bgr, kpts0_cv,
        patch_bgr, kpts1_cv,
        matches_vis, None,
        flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
    )

    out_path = os.path.join(matches_out_dir, f"{uav_imag_num:04d}_matches.png")
    cv2.imwrite(out_path, img_vis)


    # ---------- DSM, PnP, result logging ----------
 
    # --- Compute scale from resized patch back to original map patch ---
    scale_x = patch_bgr.shape[1] / img2_gray.shape[1]
    scale_y = patch_bgr.shape[0] / img2_gray.shape[0]

    obj_pts = []
    img_pts = []

    with rasterio.open(DSM_PATH) as dsm:
        dsm_band = dsm.read(1)
        nodata = dsm.nodatavals[0]

        for i in range(len(mkpts0)):
            # mkpts0: UAV image points (already scaled)
            # mkpts1: map patch points (already homography filtered)

            u_pt = mkpts0[i]

            # Scale mkpts1 to patch coordinates
            x_map = mkpts1[i,0] * scale_x
            y_map = mkpts1[i,1] * scale_y

            col_ = int(c1 + x_map)
            row_ = int(r1 + y_map)

            # Check bounds
            if row_ < 0 or row_ >= image_map_bgr.shape[0] or col_ < 0 or col_ >= image_map_bgr.shape[1]:
                continue

            try:
                e_map, n_map = px2utm_transform * (col_, row_)
                r_dsm, c_dsm = dsm.index(e_map, n_map)
                z = float(dsm_band[r_dsm, c_dsm])
                if not np.isnan(z) and z != nodata:
                    obj_pts.append([e_map, n_map, z])
                    img_pts.append([u_pt[0], u_pt[1]])
            except:
                continue

    obj_pts = np.array(obj_pts, dtype=np.float64)
    img_pts = np.array(img_pts, dtype=np.float32)

    print(f"Number of 3D points: {len(obj_pts)}")


    # check enough 3d-2d correspondences for pose estimation
    if len(obj_pts) < 6:
        print("Too few 3D points")
        continue

    # reference camera position as local origin
    camera_pos_map_expected = np.array([uav_e_map,uav_n_map,uav_alt_gps])

    # convert world coordinates into local camera-centered coordinates
    obj_pts_local = (obj_pts - camera_pos_map_expected).astype(np.float32)

    # ---------- ANCHOR PNP ----------
    # fixed initial orientation guess 
    R_anchor = np.array([[0,1,0],[1,0,0],[0,0,-1]],dtype=np.float64)

    # convert rotation matrix to rotation vector for opencv pnp
    rvec_prior,_ = cv2.Rodrigues(R_anchor)

    # ransac-based pose estimation using anchor prior
    retval,rvec_anchor,tvec_anchor,inliers_anchor = cv2.solvePnPRansac(
        obj_pts_local,img_pts,K.astype(np.float32),dist_coeffs.astype(np.float32),
        rvec=rvec_prior,tvec=np.zeros((3,1)),useExtrinsicGuess=True,
        reprojectionError=3.0,iterationsCount=10000
    )

    # stop if no inliers found
    if inliers_anchor is None:
        print("Anchor PnP failed")
        continue

    # keep only inlier correspondences
    idx = inliers_anchor.ravel()

    # refine pose using only inliers
    ok,rvec_anchor_ref,tvec_anchor_ref = cv2.solvePnP(
        obj_pts_local[idx],img_pts[idx],K.astype(np.float32),dist_coeffs.astype(np.float32),
        rvec=rvec_anchor,tvec=tvec_anchor,useExtrinsicGuess=True
    )

    # convert rotation vector back to matrix
    R_map2cam,_ = cv2.Rodrigues(rvec_anchor_ref)

    # invert to get camera-to-map rotation
    R_cam2map = R_map2cam.T

    # recover camera position in map frame
    camera_pos_anchor = (-R_cam2map @ tvec_anchor_ref).reshape(3) + camera_pos_map_expected

    # fixed boresight correction between camera and imu frame
    R_boresight = np.array([[0,1,0],[-1,0,0],[0,0,-1]],dtype=np.float64)

    # final orientation in map frame
    R_final_anchor = R_cam2map @ R_boresight

    # convert rotation matrix to yaw/pitch/roll
    v_ypr_anchor = get_euler(R_final_anchor)

    # ---------- PRIOR PNP ----------
    # convert imu quaternion to rotation matrix
    R_imu = R_from_quaternion(qw,qx,qy,qz)

    # coordinate system conversion (ned to enu style swap)
    R_ned2enu = np.array([[0,1,0],[1,0,0],[0,0,-1]],dtype=np.float64)

    # initial camera-to-map rotation from imu
    R_cam2map_prior = (R_ned2enu @ R_imu).T

    # convert to rotation vector for opencv
    rvec_guess,_ = cv2.Rodrigues(R_cam2map_prior)

    # ransac pose estimation using imu prior instead of anchor guess
    retval,rvec_prior_pnp,tvec_prior_pnp,inliers_prior = cv2.solvePnPRansac(
        obj_pts_local,img_pts,K.astype(np.float32),dist_coeffs.astype(np.float32),
        rvec=rvec_guess,tvec=np.zeros((3,1)),useExtrinsicGuess=True,
        reprojectionError=3.0,iterationsCount=10000
    )

    # check if pose estimation succeeded
    if inliers_prior is None:
        print("Prior PnP failed")
        continue

    # select inlier points
    idx = inliers_prior.ravel()

    # refine pose with inliers only
    ok,rvec_prior_ref,tvec_prior_ref = cv2.solvePnP(
        obj_pts_local[idx],img_pts[idx],K.astype(np.float32),dist_coeffs.astype(np.float32),
        rvec=rvec_prior_pnp,tvec=tvec_prior_pnp,useExtrinsicGuess=True
    )

    # convert to rotation matrix
    R_map2cam,_ = cv2.Rodrigues(rvec_prior_ref)

    # invert to camera-to-map frame
    R_cam2map = R_map2cam.T

    # compute camera position in map frame
    camera_pos_prior = (-R_cam2map @ tvec_prior_ref).reshape(3) + camera_pos_map_expected

    # final orientation with boresight correction
    R_final_prior = R_cam2map @ R_boresight

    # convert to euler angles
    v_ypr_prior = get_euler(R_final_prior)

    # ---------- SAVE RESULTS ----------
    results.append({
        # Matching
        "kp_uav": len(kpts0),
        "kp_patch": len(kpts1),
        "raw_matches": len(matches0),
        "good_matches": len(mkpts0),
        "good_ratio": len(mkpts0)/len(matches0) if len(matches0)>0 else 0,
        "homography_inliers": int(maskH.sum()) if maskH is not None else 0,
        "homography_ratio": (int(maskH.sum())/len(mkpts0)) if maskH is not None and len(mkpts0)>0 else 0.0,

        # Anchor PnP
        "num_3d2d_corr": len(obj_pts_local),
        "pnp_inliers_anchor": len(inliers_anchor) if inliers_anchor is not None else 0,
        "pnp_ratio_anchor": (len(inliers_anchor)/len(obj_pts_local)) if inliers_anchor is not None and len(obj_pts_local)>0 else 0,
        "anchor_easting_est": camera_pos_anchor[0],
        "anchor_northing_est": camera_pos_anchor[1],
        "anchor_altitude_est": camera_pos_anchor[2],
        "anchor_easting_delta": camera_pos_anchor[0]-uav_e_map,
        "anchor_northing_delta": camera_pos_anchor[1]-uav_n_map,
        "anchor_altitude_delta": camera_pos_anchor[2]-uav_alt_gps,
        "anchor_yaw_est": v_ypr_anchor[0],
        "anchor_pitch_est": v_ypr_anchor[1],
        "anchor_roll_est": v_ypr_anchor[2],
        "anchor_yaw_delta": (v_ypr_anchor[0]-np.degrees(uav_yaw)+180)%360-180,
        "anchor_pitch_delta": (v_ypr_anchor[1]-np.degrees(uav_pitch)+180)%360-180,
        "anchor_roll_delta": (v_ypr_anchor[2]-np.degrees(uav_roll)+180)%360-180,

        # Prior PnP
        "pnp_inliers_prior": len(inliers_prior) if inliers_prior is not None else 0,
        "pnp_ratio_prior": (len(inliers_prior)/len(obj_pts_local)) if inliers_prior is not None and len(obj_pts_local)>0 else 0,
        "prior_easting_est": camera_pos_prior[0],
        "prior_northing_est": camera_pos_prior[1],
        "prior_altitude_est": camera_pos_prior[2],
        "prior_easting_delta": camera_pos_prior[0]-uav_e_map,
        "prior_northing_delta": camera_pos_prior[1]-uav_n_map,
        "prior_altitude_delta": camera_pos_prior[2]-uav_alt_gps,
        "prior_yaw_est": v_ypr_prior[0],
        "prior_pitch_est": v_ypr_prior[1],
        "prior_roll_est": v_ypr_prior[2],
        "prior_yaw_delta": (v_ypr_prior[0]-np.degrees(uav_yaw)+180)%360-180,
        "prior_pitch_delta": (v_ypr_prior[1]-np.degrees(uav_pitch)+180)%360-180,
        "prior_roll_delta": (v_ypr_prior[2]-np.degrees(uav_roll)+180)%360-180,

        # IMU+GPS
        "gps_easting": uav_e_map,
        "gps_northing": uav_n_map,
        "gps_altitude": uav_alt_gps,
        "imu_yaw": np.degrees(uav_yaw),
        "imu_pitch": np.degrees(uav_pitch),
        "imu_roll": np.degrees(uav_roll),

        # ID
        "image_id": uav_imag_num,
        "filename": filename
    })

    torch.cuda.empty_cache()

# ---------- EXPORT ----------
results_df = pd.DataFrame(results)
results_df.to_csv("uav_pose_results_pt_trained_BF.csv", index=False)
print("Saved results to uav_pose_results_pt_trained_BF.csv")
```
