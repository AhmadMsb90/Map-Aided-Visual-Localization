#### Imports

```python
import os
import cv2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob
import sys
import rasterio
from rasterio.transform import rowcol
from pyproj import Transformer
import math
import torch
from pathlib import Path

# -------------------- DISK + LIGHTGLUE -------------------
SG_ROOT = Path("/home/jovyan/Models/LightGlue")
sys.path.insert(0, str(SG_ROOT))

from lightglue.lightglue import LightGlue
from lightglue.disk import DISK
```

#### Parameters

```python
MIN_GOOD_MATCHES = 8  # min matches required
RANSAC_REPROJ_THRESHOLD = 5.0    # Homography RANSAC threshold (pixels)
MAX_DIM = 1024 # max image dimension for resizing
cv2.setRNGSeed(0)
np.random.seed(0)

# -------------------- DISK + LightGlue --------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

disk = DISK(
    detection_threshold=0.6,
    max_num_keypoints=4000
).to(device).eval()

matcher = LightGlue(
    features="disk",
    pruning_keypoint_thresholds=0.2
).to(device).eval()
```

#### Reading Orthomosaic Map

```python
#-------------------- PATH --------------------
ORTHO_PATH = "../shared/datasets/garrulus-arnsberg-forest/data_collection/2025-11-19_field-D-E/post-processing-results/2025-11-19_field-D_flight-12_rgb-w-gcp/rgb_wide_images-only_orthomosaic.tif"
DSM_PATH = "../shared/datasets/garrulus-arnsberg-forest/data_collection/2025-11-19_field-D-E/post-processing-results/2025-11-19_field-D_flight-10_rgb-w-gcp/rgb_wide_images-only_dtm.tif"
# UAV image resolution in pixels (width x height)
uav_image_res_w = 1936
uav_image_res_h = 1460

# UAV GSD in m/px
gsd_uav_x_m = 0.6 / 100.0
gsd_uav_y_m = 0.6 / 100.0

#-------------------- CHECK ORTHOMOSAIC FILE --------------------
if not os.path.exists(ORTHO_PATH):
    raise FileNotFoundError(f"Orthomosaic not found: {ORTHO_PATH}")

#-------------------- OPEN ORTHOMOSAIC AND READ DATA --------------------
with rasterio.open(ORTHO_PATH) as ds:
    # Read the entire orthomosaic image 
    image_map_raw = ds.read(out_dtype="uint8")
    # get affine transform to convert pixel to map coordinates (UTM)
    px2utm_transform = ds.transform
    # get the Coordinate Reference System (CRS) of the orthomosaic
    ortho_crs = ds.crs
    # get image spatial bounds
    bounds = ds.bounds

    print("Orthomosaic CRS:", ortho_crs)
    print("Orthomosaic size (W x H):", ds.width, "x", ds.height)

    # calculate GSD from transform matrix (pixel size in meters)
    GSD_x = abs(px2utm_transform.a)
    GSD_y = abs(px2utm_transform.e)
    print(f"Orthomosaic GSD (m/pixel): ({GSD_x:.6f}, {GSD_y:.6f})")

#-------------------- PREPROCESS IMAGE BANDS --------------------
# keep the RGB bounds
if image_map_raw.shape[0] > 3:
    image_map_raw = image_map_raw[:3, :, :]

# convert image array to height x width x bands format for OpenCV
image_map_rgb = np.transpose(image_map_raw, (1, 2, 0))
# convert from RGB to BGR 
image_map_bgr = cv2.cvtColor(image_map_rgb, cv2.COLOR_RGB2BGR)

# get orthomosaic image dimensions
h_map, w_map = image_map_bgr.shape[:2]

#-------------------- TRANSFORM UAV GPS COORDINATES TO ORTHOMOSAIC CRS --------------------
# create transformer from WGS84 to orthomosaic CRS
transformer = Transformer.from_crs("EPSG:4326", ortho_crs, always_xy=True)
```

#### Helper Functions

```python
def to_gray(bgr):
    # convert color image to grayscale
    return cv2.cvtColor(bgr, cv2.COLOR_BGR2GRAY)

def resize_keep(img, max_dim=MAX_DIM):
    # resize image if too large, keep same aspect ratio
    h, w = img.shape[:2]
    
    scale = 1.0
    if max(h, w) > max_dim:
        scale = max_dim / float(max(h, w))
        img = cv2.resize(img, (int(w*scale), int(h*scale)), interpolation=cv2.INTER_AREA)
    return img, scale

def get_euler(R):
    # extract euler angles (in degrees) from rotation matrix
    sy = np.sqrt(R[0,0]**2 + R[1,0]**2)
    if sy > 1e-6:
        return np.degrees([
            np.arctan2(R[1,0], R[0,0]),   # yaw
            np.arctan2(-R[2,0], sy),      # pitch
            np.arctan2(R[2,1], R[2,2])    # roll
        ])
    # fallback when near singular case
    return np.degrees([
        0,
        np.arctan2(-R[2,0], sy),
        np.arctan2(-R[1,2], R[1,1])
    ])


def R_from_quaternion(w, x, y, z):
    # normalize quaternion first
    n = np.sqrt(w*w + x*x + y*y + z*z)
    w, x, y, z = w/n, x/n, y/n, z/n

    # build rotation matrix from quaternion
    return np.array([
        [1 - 2*(y*y + z*z),     2*(x*y - z*w),     2*(x*z + y*w)],
        [    2*(x*y + z*w), 1 - 2*(x*x + z*z),     2*(y*z - x*w)],
        [    2*(x*z - y*w),     2*(y*z + x*w), 1 - 2*(x*x + y*y)]
    ], dtype=np.float64)


def compute_reprojection_error(obj_pts, img_pts, rvec, tvec, K, dist):
    # compute reprojection error (mean + std)
    proj_pts, _ = cv2.projectPoints(obj_pts, rvec, tvec, K, dist)
    proj_pts = proj_pts.reshape(-1, 2)
    img_pts = img_pts.reshape(-1, 2)
    errors = np.linalg.norm(proj_pts - img_pts, axis=1)
    return errors.mean(), errors.std()
```

#### Camera Intrinsics

```python
# camera intrinsic matrix (focal length + principal point)
K = np.array([
    [1806.84457,     0.0,        945.416679],
    [     0.0,    1806.43529,    738.031572],
    [     0.0,        0.0,          1.0     ]
], dtype=np.float64)

# lens distortion parameters from calibration
dist_coeffs = np.array([
    -0.20484306, 
     0.1278626, 
    -0.00041228, 
     0.0005107, 
    -0.06076351
]).reshape(-1, 1)
```

#### Pose Estimation Pipeline

```python
# ---------- LOOP OVER IMAGES ----------
image_dir = "../Field_D_13_images/extracted_images2/"
meta_path = "../Field_D_13_images/extracted_images2/metadata_with_euler.csv"

meta_data = pd.read_csv(meta_path)
image_paths = sorted(glob.glob(os.path.join(image_dir, "frame_*.png")))

results = []

for image_path in image_paths:

    filename = os.path.basename(image_path)
    uav_imag_num = int(filename.split("_")[1].split(".")[0])

    if uav_imag_num < 214:
        continue

    print("\nProcessing:", uav_imag_num, filename)

    # ---------- LOAD METADATA ----------
    row = meta_data.iloc[uav_imag_num]

    uav_lat_gps = float(row["gps_latitude"])
    uav_lon_gps = float(row["gps_longitude"])
    uav_alt_gps = float(row["gps_altitude"])

    uav_roll = float(row["roll"])
    uav_pitch = float(row["pitch"])
    uav_yaw = float(row["yaw"])

    qx = float(row["imu_orientation_x"])
    qy = float(row["imu_orientation_y"])
    qz = float(row["imu_orientation_z"])
    qw = float(row["imu_orientation_w"])

    # ---------- LOAD UAV IMAGE ----------
    img_uav_bgr = cv2.imread(image_path)

    if img_uav_bgr is None:
        print("Image read failed")
        continue

    img_uav_bgr = img_uav_bgr.astype("uint8")

    # ---------- ROTATE UAV IMAGE ----------
    yaw_deg = np.rad2deg(uav_yaw)
    rotation_angle = -90.0 + yaw_deg

    h, w = img_uav_bgr.shape[:2]
    center = (w/2.0, h/2.0)

    M = cv2.getRotationMatrix2D(center, rotation_angle, 1.0)

    cos = abs(M[0,0])
    sin = abs(M[0,1])

    new_w = int(h*sin + w*cos)
    new_h = int(h*cos + w*sin)

    M[0,2] += (new_w - w)/2.0
    M[1,2] += (new_h - h)/2.0

    img_uav_rot_bgr = cv2.warpAffine(
        img_uav_bgr,
        M,
        (new_w,new_h),
        flags=cv2.INTER_LINEAR,
        borderMode=cv2.BORDER_CONSTANT
    )

    # ---------- MAP COORDINATES ----------
    uav_e_map, uav_n_map = transformer.transform(uav_lon_gps, uav_lat_gps)

    row_f, col_f = rowcol(px2utm_transform, uav_e_map, uav_n_map)

    r = int(round(row_f))
    c = int(round(col_f))

    # ---------- PATCH EXTRACTION ----------
    ortho_per_uav_x = gsd_uav_x_m / GSD_x
    ortho_per_uav_y = gsd_uav_y_m / GSD_y

    crop_w = int(uav_image_res_w * ortho_per_uav_x) // 2
    crop_h = int(uav_image_res_h * ortho_per_uav_y) // 2

    r1 = max(r - crop_h, 0)
    r2 = min(r + crop_h, h_map)

    c1 = max(c - crop_w, 0)
    c2 = min(c + crop_w, w_map)

    patch_bgr = image_map_bgr[r1:r2, c1:c2].copy()

    if patch_bgr.size == 0:
        print("Patch extraction failed")
        continue

    # ---------- RESIZE ----------
    img1_rs, s1 = resize_keep(img_uav_rot_bgr)
    img2_rs, s2 = resize_keep(patch_bgr)

    img1_gray = to_gray(img1_rs)
    img2_gray = to_gray(img2_rs)

    # ---------- DISK + LIGHTGLUE ----------
    img0_tensor = torch.from_numpy(img1_gray.astype(np.float32) / 255.0)[None, None].to(device)
    img1_tensor = torch.from_numpy(img2_gray.astype(np.float32) / 255.0)[None, None].to(device)

    with torch.no_grad():

        feats0 = disk({'image': img0_tensor})
        feats1 = disk({'image': img1_tensor})

        match_input = {
            'image0': {
                'keypoints': feats0['keypoints'],
                'descriptors': feats0['descriptors'],
                'image': img0_tensor,
            },
            'image1': {
                'keypoints': feats1['keypoints'],
                'descriptors': feats1['descriptors'],
                'image': img1_tensor,
            }
        }

        match_output = matcher(match_input)

    kpts0 = feats0['keypoints'][0].cpu().numpy()
    kpts1 = feats1['keypoints'][0].cpu().numpy()

    matches0 = match_output['matches0'][0].cpu().numpy()
    conf = match_output['matching_scores0'][0].cpu().numpy()

    valid = matches0 != -1

    mkpts0 = kpts0[valid]
    mkpts1 = kpts1[matches0[valid]]

    raw_matches = len(kpts0)
    good_matches = len(mkpts0)

    print("LightGlue matches:", good_matches)

    if good_matches < 20:
        print("Too few matches")
        continue






    # ---------- HOMOGRAPHY FILTER ----------
    src = mkpts0.reshape(-1,1,2)
    dst = mkpts1.reshape(-1,1,2)

    H, mask = cv2.findHomography(src, dst, cv2.RANSAC, 5.0)

    if H is None:
        print("Homography failed")
        continue

    mask = mask.ravel().astype(bool)

    mkpts0 = mkpts0[mask]
    mkpts1 = mkpts1[mask]

    # ---------- RESCALE AND UNDO ROTATION (BACK TO ORIGINAL UAV IMAGE FRAME) ----------
    # scale back from resized coordinates to full-rotated coordinates
    mkpts0_full_rot = mkpts0 / s1

    # invert rotation back to the original raw camera frame
    H_inv = cv2.invertAffineTransform(M)
    mkpts0_original = cv2.transform(mkpts0_full_rot.reshape(-1, 1, 2), H_inv).reshape(-1, 2)


    # ---------- VISUALIZE AND SAVE MATCHES ----------
    img_uav_vis = img1_rs.copy()
    patch_vis = img2_rs.copy()

    # Ensure both are 3-channel
    if len(img_uav_vis.shape) == 2:
        img_uav_vis = cv2.cvtColor(img_uav_vis, cv2.COLOR_GRAY2BGR)
    if len(patch_vis.shape) == 2:
        patch_vis = cv2.cvtColor(patch_vis, cv2.COLOR_GRAY2BGR)

    # Create side-by-side canvas
    h1, w1 = img_uav_vis.shape[:2]
    h2, w2 = patch_vis.shape[:2]

    canvas = np.zeros((max(h1, h2), w1 + w2, 3), dtype=np.uint8)

    canvas[:h1, :w1] = img_uav_vis
    canvas[:h2, w1:w1+w2] = patch_vis

    # Draw matches
    for pt0, pt1 in zip(mkpts0, mkpts1):
        x0, y0 = int(pt0[0]), int(pt0[1])
        x1, y1 = int(pt1[0] + w1), int(pt1[1])

        color = tuple(np.random.randint(0, 255, 3).tolist())

        cv2.circle(canvas, (x0, y0), 3, color, -1)
        cv2.circle(canvas, (x1, y1), 3, color, -1)
        cv2.line(canvas, (x0, y0), (x1, y1), color, 1)

    # Save
    vis_dir = "./matches_out_dsk_lg_both_default/"
    os.makedirs(vis_dir, exist_ok=True)

    save_path = os.path.join(vis_dir, f"match_{uav_imag_num}.png")
    cv2.imwrite(save_path, canvas)
    print("Saved match image:", save_path)


    
    if len(mkpts0) >= 4:
        src_pts = mkpts0.reshape(-1,1,2)
        dst_pts = mkpts1.reshape(-1,1,2)
        H, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, 3.0)
        maskH = mask.ravel().astype(bool)
        homography_inliers = int(maskH.sum())
        homography_ratio = homography_inliers / len(mkpts0)
    else:
        homography_inliers = 0
        homography_ratio = 0.0

    # ---------- DSM SAMPLING ----------
    obj_pts = []
    img_pts = []
    
    # Scale back patch keypoints from resized (img2_rs) to full patch size (patch_bgr)
    mkpts1_full_patch = mkpts1 / s2

    with rasterio.open(DSM_PATH) as dsm:
        dsm_band = dsm.read(1)
        nodata = dsm.nodatavals[0]

        for i in range(len(mkpts0_original)):
            # Use the scaled-back patch coordinates
            col_patch_full = mkpts1_full_patch[i][0]
            row_patch_full = mkpts1_full_patch[i][1]

            # Use patch top-left (c1, r1) to find global map pixel
            col_map = int(c1 + col_patch_full)
            row_map = int(r1 + row_patch_full)

            try:
                # Pixel to UTM
                e_map, n_map = px2utm_transform * (col_map, row_map)
                
                # Sample Elevation
                r_dsm, c_dsm = dsm.index(e_map, n_map)
                z = float(dsm_band[r_dsm, c_dsm])

                if not np.isnan(z) and z != nodata:
                    obj_pts.append([e_map, n_map, z])
                    # Use the UNROTATED, FULL-SCALE UAV pixel coordinates
                    img_pts.append([mkpts0_original[i][0], mkpts0_original[i][1]])
            except:
                continue



    obj_pts = np.array(obj_pts, dtype=np.float64)
    img_pts = np.array(img_pts, dtype=np.float32)

    # skip frame if too few 3d points available for reliable pose estimation
    if len(obj_pts) < 6:
        print("Too few 3D points")
        continue

    # reference camera position in map coordinates
    camera_pos_map_expected = np.array([uav_e_map,uav_n_map,uav_alt_gps])

    # shift 3d world points into local coordinate system centered at camera
    obj_pts_local = (obj_pts - camera_pos_map_expected).astype(np.float32)

    # ---------- anchor pnp ----------
    # fixed initial rotation guess for pose estimation
    R_anchor = np.array([[0,1,0],[1,0,0],[0,0,-1]],dtype=np.float64)

    # convert rotation matrix to rotation vector for opencv
    rvec_prior,_ = cv2.Rodrigues(R_anchor)

    # estimate pose using ransac with prior rotation and zero translation
    retval,rvec_anchor,tvec_anchor,inliers_anchor = cv2.solvePnPRansac(
        obj_pts_local,img_pts,
        K.astype(np.float32),
        dist_coeffs.astype(np.float32),
        rvec=rvec_prior,
        tvec=np.zeros((3,1)),
        useExtrinsicGuess=True,
        reprojectionError=3.0,
        iterationsCount=10000
    )

    # skip frame if anchor pnp fails
    if inliers_anchor is None:
        print("Anchor PnP failed")
        continue

    # keep only inlier correspondences
    idx = inliers_anchor.ravel()

    # refine pose using only inlier points
    ok,rvec_anchor_ref,tvec_anchor_ref = cv2.solvePnP(
        obj_pts_local[idx],
        img_pts[idx],
        K.astype(np.float32),
        dist_coeffs.astype(np.float32),
        rvec=rvec_anchor,
        tvec=tvec_anchor,
        useExtrinsicGuess=True
    )

    # convert rotation vector to rotation matrix
    R_map2cam,_ = cv2.Rodrigues(rvec_anchor_ref)

    # convert to camera-to-map rotation
    R_cam2map = R_map2cam.T

    # estimate camera position in map coordinates
    camera_pos_anchor = (-R_cam2map @ tvec_anchor_ref).reshape(3) + camera_pos_map_expected

    # boresight correction 
    R_boresight = np.array([[0,1,0],[-1,0,0],[0,0,-1]],dtype=np.float64)

    # final corrected orientation in map frame
    R_final_anchor = R_cam2map @ R_boresight

    # convert rotation matrix to yaw pitch roll angles
    v_ypr_anchor = get_euler(R_final_anchor)

    # compute reprojection error for anchor solution quality
    mean_err_anchor, std_err_anchor = compute_reprojection_error(
    obj_pts_local[idx],
    img_pts[idx],
    rvec_anchor_ref,
    tvec_anchor_ref,
    K.astype(np.float32),
    dist_coeffs.astype(np.float32)
    )

    # ---------- prior pnp ----------
    # compute imu-based orientation from quaternion
    R_imu = R_from_quaternion(qw,qx,qy,qz)

    # convert navigation frame 
    R_ned2enu = np.array([[0,1,0],[1,0,0],[0,0,-1]],dtype=np.float64)

    # convert imu orientation into camera-to-map rotation estimate
    R_cam2map_prior = (R_ned2enu @ R_imu).T

    # convert rotation matrix to rotation vector 
    rvec_guess,_ = cv2.Rodrigues(R_cam2map_prior)

    # initial translation guess set to zero
    tvec_guess = np.zeros((3,1))

    # solve pnp using imu-based initialization
    retval,rvec_prior_pnp,tvec_prior_pnp,inliers_prior = cv2.solvePnPRansac(
        obj_pts_local,img_pts,
        K.astype(np.float32),
        dist_coeffs.astype(np.float32),
        rvec=rvec_guess,
        tvec=tvec_guess,
        useExtrinsicGuess=True,
        reprojectionError=3.0,
        iterationsCount=10000
    )

    # skip frame if prior pnp fails
    if inliers_prior is None:
        print("Prior PnP failed")
        continue

    # extract inlier indices
    idx = inliers_prior.ravel()

    # refine pose using inlier set
    ok,rvec_prior_ref,tvec_prior_ref = cv2.solvePnP(
        obj_pts_local[idx],
        img_pts[idx],
        K.astype(np.float32),
        dist_coeffs.astype(np.float32),
        rvec=rvec_prior_pnp,
        tvec=tvec_prior_pnp,
        useExtrinsicGuess=True
    )

    # convert rotation vector to matrix
    R_map2cam,_ = cv2.Rodrigues(rvec_prior_ref)

    # convert to camera-to-map frame
    R_cam2map = R_map2cam.T

    # estimate camera position in map coordinates
    camera_pos_prior = (-R_cam2map @ tvec_prior_ref).reshape(3) + camera_pos_map_expected

    # apply boresight correction
    R_final_prior = R_cam2map @ R_boresight

    # convert final rotation to yaw pitch roll
    v_ypr_prior = get_euler(R_final_prior)

    # compute reprojection error for prior solution
    mean_err_prior, std_err_prior = compute_reprojection_error(
    obj_pts_local[idx],
    img_pts[idx],
    rvec_prior_ref,
    tvec_prior_ref,
    K.astype(np.float32),
    dist_coeffs.astype(np.float32)
    )



    # ---------- SAVE RESULTS ----------
    results.append({
        # Matching
        "kp_uav": len(kpts0),
        "kp_patch": len(kpts1),
        "raw_matches": len(matches0),
        "good_matches": len(mkpts0),
        "good_ratio": len(mkpts0)/len(matches0) if len(matches0)>0 else 0,
        "homography_inliers": int(maskH.sum()) if maskH is not None else 0,
        "homography_ratio": (int(maskH.sum())/len(mkpts0)) if maskH is not None and len(mkpts0)>0 else 0.0,

        # Anchor PnP
        "num_3d2d_corr": len(obj_pts_local),
        "pnp_inliers_anchor": len(inliers_anchor) if inliers_anchor is not None else 0,
        "pnp_ratio_anchor": (len(inliers_anchor)/len(obj_pts_local)) if inliers_anchor is not None and len(obj_pts_local)>0 else 0,
        "anchor_easting_est": camera_pos_anchor[0],
        "anchor_northing_est": camera_pos_anchor[1],
        "anchor_altitude_est": camera_pos_anchor[2],
        "anchor_easting_delta": camera_pos_anchor[0]-uav_e_map,
        "anchor_northing_delta": camera_pos_anchor[1]-uav_n_map,
        "anchor_altitude_delta": camera_pos_anchor[2]-uav_alt_gps,
        "anchor_yaw_est": v_ypr_anchor[0],
        "anchor_pitch_est": v_ypr_anchor[1],
        "anchor_roll_est": v_ypr_anchor[2],
        "anchor_yaw_delta": (v_ypr_anchor[0]-np.degrees(uav_yaw)+180)%360-180,
        "anchor_pitch_delta": (v_ypr_anchor[1]-np.degrees(uav_pitch)+180)%360-180,
        "anchor_roll_delta": (v_ypr_anchor[2]-np.degrees(uav_roll)+180)%360-180,
        "anchor_reproj_mean": mean_err_anchor,
        "anchor_reproj_std": std_err_anchor,

        # Prior PnP
        "pnp_inliers_prior": len(inliers_prior) if inliers_prior is not None else 0,
        "pnp_ratio_prior": (len(inliers_prior)/len(obj_pts_local)) if inliers_prior is not None and len(obj_pts_local)>0 else 0,
        "prior_easting_est": camera_pos_prior[0],
        "prior_northing_est": camera_pos_prior[1],
        "prior_altitude_est": camera_pos_prior[2],
        "prior_easting_delta": camera_pos_prior[0]-uav_e_map,
        "prior_northing_delta": camera_pos_prior[1]-uav_n_map,
        "prior_altitude_delta": camera_pos_prior[2]-uav_alt_gps,
        "prior_yaw_est": v_ypr_prior[0],
        "prior_pitch_est": v_ypr_prior[1],
        "prior_roll_est": v_ypr_prior[2],
        "prior_yaw_delta": (v_ypr_prior[0]-np.degrees(uav_yaw)+180)%360-180,
        "prior_pitch_delta": (v_ypr_prior[1]-np.degrees(uav_pitch)+180)%360-180,
        "prior_roll_delta": (v_ypr_prior[2]-np.degrees(uav_roll)+180)%360-180,
        "prior_reproj_mean": mean_err_prior,
        "prior_reproj_std": std_err_prior,

        # IMU+GPS
        "gps_easting": uav_e_map,
        "gps_northing": uav_n_map,
        "gps_altitude": uav_alt_gps,
        "imu_yaw": np.degrees(uav_yaw),
        "imu_pitch": np.degrees(uav_pitch),
        "imu_roll": np.degrees(uav_roll),

        # ID
        "image_id": uav_imag_num,
        "filename": filename
    })
    del img0_tensor, img1_tensor, feats0, feats1, match_output
    torch.cuda.empty_cache()

# ---------- EXPORT ----------
results_df = pd.DataFrame(results)

results_df.to_csv("uav_pose_disk_lihtglue.csv", index=False)

print("Saved results to uav_pose_disk_lihtglue.csv")
```

#### Results

```python


import os
import pandas as pd
import numpy as np
import rasterio
from pyproj import Transformer
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# -------------------- CONFIG --------------------
results_csv = "uav_pose_disk_lihtglue.csv"
meta_path = "../Field_D_13_images/extracted_images2/metadata_with_euler.csv"
DSM_PATH = "../shared/datasets/garrulus-arnsberg-forest/data_collection/2025-11-19_field-D-E/post-processing-results/2025-11-19_field-D_flight-10_rgb-w-gcp/rgb_wide_images-only_dtm.tif"

height_bins = [10, 11.7, 13.3, 15]
height_labels = ["low", "medium", "high"]

# =========================================================
# LOAD DATA
# =========================================================
df = pd.read_csv(results_csv)
meta = pd.read_csv(meta_path)

# =========================================================
# COLUMN FETCH
# =========================================================
def get_required_column(df, possible_names, label):
    for name in possible_names:
        if name in df.columns:
            print(f"[OK] Using column for {label}: {name}")
            return df[name]
    raise KeyError(f"[ERROR] None found for {label}: {possible_names}")

# =========================================================
# DERIVED METRICS
# =========================================================
df["anchor_pnp_inlier_ratio"] = 100 * df["pnp_ratio_anchor"]
df["prior_pnp_inlier_ratio"]  = 100 * df["pnp_ratio_prior"]

df["anchor_reproj"] = get_required_column(df, [
    "anchor_reproj_mean",
    "reproj_error_anchor_mean",
    "anchor_reprojection_error"
], "anchor reprojection")

df["prior_reproj"] = get_required_column(df, [
    "prior_reproj_mean",
    "reproj_error_prior_mean",
    "prior_reprojection_error"
], "prior reprojection")

# =========================================================
# AGL COMPUTATION
# =========================================================
agl_list = []

with rasterio.open(DSM_PATH) as src:
    dsm = src.read(1)
    nodata = src.nodatavals[0]

    transformer = Transformer.from_crs("EPSG:4326", src.crs, always_xy=True)

    n_rows, n_cols = dsm.shape
    bounds = src.bounds

    for i in range(len(df)):
        row = meta.iloc[i]

        lon, lat, alt = float(row["gps_longitude"]), float(row["gps_latitude"]), float(row["gps_altitude"])
        e, n = transformer.transform(lon, lat)

        if not (bounds.left <= e <= bounds.right and bounds.bottom <= n <= bounds.top):
            agl_list.append(np.nan)
            continue

        r, c = src.index(e, n)
        r, c = np.clip(r, 0, n_rows - 1), np.clip(c, 0, n_cols - 1)

        ground = float(dsm[r, c])

        if nodata is not None and ground == nodata:
            agl_list.append(np.nan)
        else:
            agl_list.append(alt - ground)

df["agl"] = agl_list
df["height_class"] = pd.cut(df["agl"], bins=height_bins, labels=height_labels)

# =========================================================
# POSITION ERRORS 
# =========================================================
df["anchor_easting_error"]  = np.abs(df["anchor_easting_delta"])
df["anchor_northing_error"] = np.abs(df["anchor_northing_delta"])
df["anchor_altitude_error"] = np.abs(df["anchor_altitude_delta"])

df["anchor_pos_error"] = np.sqrt(
    df["anchor_easting_delta"]**2 +
    df["anchor_northing_delta"]**2 +
    df["anchor_altitude_delta"]**2
)

df["prior_easting_error"]  = np.abs(df["prior_easting_delta"])
df["prior_northing_error"] = np.abs(df["prior_northing_delta"])
df["prior_altitude_error"] = np.abs(df["prior_altitude_delta"])

df["prior_pos_error"] = np.sqrt(
    df["prior_easting_delta"]**2 +
    df["prior_northing_delta"]**2 +
    df["prior_altitude_delta"]**2
)

# =========================================================
# ORIENTATION ERRORS
# =========================================================
df["anchor_yaw_error"]   = np.abs(df["anchor_yaw_delta"])
df["anchor_pitch_error"] = np.abs(df["anchor_pitch_delta"])
df["anchor_roll_error"]  = np.abs(df["anchor_roll_delta"])

df["prior_yaw_error"]   = np.abs(df["prior_yaw_delta"])
df["prior_pitch_error"] = np.abs(df["prior_pitch_delta"])
df["prior_roll_error"]  = np.abs(df["prior_roll_delta"])

# =========================================================
# COMBINED ORIENTATION
# =========================================================
df["anchor_ori_error"] = np.sqrt(
    df["anchor_yaw_error"]**2 +
    df["anchor_pitch_error"]**2 +
    df["anchor_roll_error"]**2
)

df["prior_ori_error"] = np.sqrt(
    df["prior_yaw_error"]**2 +
    df["prior_pitch_error"]**2 +
    df["prior_roll_error"]**2
)

# =========================================================
# MATCHING
# =========================================================
df["matching_ratio"] = 100 * df["good_matches"] / df["raw_matches"].replace(0, np.nan)

# =========================================================
# QUANTILE FUNCTION
# =========================================================
def qtable(name, s):
    s = s.dropna()
    if len(s) == 0:
        print(f"| {name} | NaN | NaN | NaN | NaN | NaN |")
        return
    q = [0.50, 0.75, 0.90, 0.95, 0.99]
    v = s.quantile(q).values
    print(f"| {name} | {v[0]:.3f} | {v[1]:.3f} | {v[2]:.3f} | {v[3]:.3f} | {v[4]:.3f} |")

# =========================================================
# GLOBAL RESULTS 
# =========================================================
print("\n=========== GLOBAL RESULTS (FULL) ===========\n")
print("| Metric | P50 | P75 | P90 | P95 | P99 |")
print("|--------|-----|-----|-----|-----|-----|")

qtable("Matching ratio (%)", df["matching_ratio"])

qtable("Anchor PnP inlier ratio (%)", df["anchor_pnp_inlier_ratio"])
qtable("Prior PnP inlier ratio (%)", df["prior_pnp_inlier_ratio"])

qtable("Anchor position error (m)", df["anchor_pos_error"])
qtable("Prior position error (m)", df["prior_pos_error"])

qtable("Anchor easting error (m)", df["anchor_easting_error"])
qtable("Anchor northing error (m)", df["anchor_northing_error"])
qtable("Anchor altitude error (m)", df["anchor_altitude_error"])

qtable("Prior easting error (m)", df["prior_easting_error"])
qtable("Prior northing error (m)", df["prior_northing_error"])
qtable("Prior altitude error (m)", df["prior_altitude_error"])

qtable("Anchor yaw error (deg)", df["anchor_yaw_error"])
qtable("Anchor pitch error (deg)", df["anchor_pitch_error"])
qtable("Anchor roll error (deg)", df["anchor_roll_error"])

qtable("Prior yaw error (deg)", df["prior_yaw_error"])
qtable("Prior pitch error (deg)", df["prior_pitch_error"])
qtable("Prior roll error (deg)", df["prior_roll_error"])

# ➕ ADDITION (from script 2)
qtable("Anchor orientation error (deg)", df["anchor_ori_error"])
qtable("Prior orientation error (deg)", df["prior_ori_error"])

qtable("Anchor reprojection error (px)", df["anchor_reproj"])
qtable("Prior reprojection error (px)", df["prior_reproj"])

# =========================================================
# PER HEIGHT RESULTS 
# =========================================================
print("\n=========== PER HEIGHT RESULTS (COMBINED) ===========")

for h in height_labels:

    sub = df[df["height_class"] == h]

    if len(sub) == 0:
        continue

    print("\n========================================")
    print("HEIGHT:", h, "| frames:", len(sub))
    print("========================================")

    print("\n--- POSITION ERROR (COMBINED) ---")
    print("| Metric | P50 | P75 | P90 | P95 | P99 |")
    print("|--------|-----|-----|-----|-----|-----|")

    qtable("Anchor position error (m)", sub["anchor_pos_error"])
    qtable("Prior position error (m)", sub["prior_pos_error"])

    print("\n--- ORIENTATION ERROR (COMBINED) ---")
    print("| Metric | P50 | P75 | P90 | P95 | P99 |")
    print("|--------|-----|-----|-----|-----|-----|")

    qtable("Anchor orientation error (deg)", sub["anchor_ori_error"])
    qtable("Prior orientation error (deg)", sub["prior_ori_error"])




# -------------------- HISTOGRAMS --------------------
def plot_hist(series, bins=30, xlabel="", ylabel="Frames", title="", color="skyblue"):
    plt.figure(figsize=(10,4))
    plt.hist(series.dropna(), bins=bins, color=color, edgecolor="black")
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.grid(True)
    plt.show()

# Matching ratio & homography
plot_hist(df["matching_ratio"], xlabel="Matching Ratio (%)", title="Matching Ratio Distribution (DISK + LightGlue)")


# UAV AGL
min_agl = 5.0
agl_filtered = df["agl"][(df["agl"] > min_agl) & (~df["agl"].isna())]
plot_hist(agl_filtered, xlabel="AGL (m)", title="UAV AGL Distribution ", color="orange")


# KDE per height
colors={"low":"skyblue","medium":"orange","high":"green"}
plt.figure(figsize=(10,5))
x_min, x_max = df["matching_ratio"].min(), df["matching_ratio"].max()
x_range = np.linspace(x_min, x_max, 500)

for h in height_labels:
    sub = df[df["height_class"]==h]["matching_ratio"].dropna()
    if len(sub)<5: continue
    kde = gaussian_kde(sub)
    y = kde(x_range)
    y = y / y.sum() * 100  # normalize
    plt.plot(x_range, y, label=f"{h} AGL", color=colors[h], linewidth=2)

plt.xlabel("Matching Ratio (%)")
plt.ylabel("Distribution (%)")
plt.title("Matching Ratio Distribution by Height Class (DISK + LightGlue)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# -------------------- MASSIVE ERROR THRESHOLDS --------------------

pos_thresh = 5.0      # meters
yaw_thresh = 15.0     # degrees
pitch_thresh = 10.0
roll_thresh = 10.0

# -------------------- MASSIVE ERROR FLAGS --------------------

# Anchor
df["massive_anchor_e"] = np.abs(df["anchor_easting_delta"]) > pos_thresh
df["massive_anchor_n"] = np.abs(df["anchor_northing_delta"]) > pos_thresh
df["massive_anchor_z"] = np.abs(df["anchor_altitude_delta"]) > pos_thresh

df["massive_anchor_roll"]  = np.abs(df["anchor_roll_delta"])  > roll_thresh
df["massive_anchor_pitch"] = np.abs(df["anchor_pitch_delta"]) > pitch_thresh
df["massive_anchor_yaw"]   = np.abs(df["anchor_yaw_delta"])   > yaw_thresh

df["massive_anchor_total_pos"] = df["anchor_pos_error"] > pos_thresh


# Prior
df["massive_prior_e"] = np.abs(df["prior_easting_delta"]) > pos_thresh
df["massive_prior_n"] = np.abs(df["prior_northing_delta"]) > pos_thresh
df["massive_prior_z"] = np.abs(df["prior_altitude_delta"]) > pos_thresh

df["massive_prior_roll"]  = np.abs(df["prior_roll_delta"])  > roll_thresh
df["massive_prior_pitch"] = np.abs(df["prior_pitch_delta"]) > pitch_thresh
df["massive_prior_yaw"]   = np.abs(df["prior_yaw_delta"])   > yaw_thresh

df["massive_prior_total_pos"] = df["prior_pos_error"] > pos_thresh


print("\n=========== MASSIVE ERROR ANALYSIS PER HEIGHT CLASS ===========")

for h in height_labels:

    sub = df[df["height_class"] == h]

    if len(sub) == 0:
        continue

    n = len(sub)

    print("\n====================================================")
    print("HEIGHT CLASS:", h)
    print("Frames:", n)

    # ---------------- ANCHOR ----------------

    print("\n--- Anchor Solution ---")

    print("Easting massive error % :", 100 * sub["massive_anchor_e"].sum() / n)
    print("Northing massive error %:", 100 * sub["massive_anchor_n"].sum() / n)
    print("Altitude massive error %:", 100 * sub["massive_anchor_z"].sum() / n)

    print("Roll massive error % :", 100 * sub["massive_anchor_roll"].sum() / n)
    print("Pitch massive error %:", 100 * sub["massive_anchor_pitch"].sum() / n)
    print("Yaw massive error %  :", 100 * sub["massive_anchor_yaw"].sum() / n)

    print("Total position massive error %:",
          100 * sub["massive_anchor_total_pos"].sum() / n)


    # ---------------- PRIOR ----------------

    print("\n--- Prior Solution ---")

    print("Easting massive error % :", 100 * sub["massive_prior_e"].sum() / n)
    print("Northing massive error %:", 100 * sub["massive_prior_n"].sum() / n)
    print("Altitude massive error %:", 100 * sub["massive_prior_z"].sum() / n)

    print("Roll massive error % :", 100 * sub["massive_prior_roll"].sum() / n)
    print("Pitch massive error %:", 100 * sub["massive_prior_pitch"].sum() / n)
    print("Yaw massive error %  :", 100 * sub["massive_prior_yaw"].sum() / n)

    print("Total position massive error %:",
          100 * sub["massive_prior_total_pos"].sum() / n)
```
