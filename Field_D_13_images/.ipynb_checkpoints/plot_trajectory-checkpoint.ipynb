```python
import os
import cv2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pyproj import Transformer
import rasterio

#-------------------- PATH --------------------
DSM_PATH = "../shared/datasets/garrulus-arnsberg-forest/data_collection/2025-11-19_field-D-E/post-processing-results/2025-11-19_field-D_flight-10_rgb-w-gcp/rgb_wide_images-only_dtm.tif"

#-------------------- CHECK DSM FILE --------------------
if not os.path.exists(DSM_PATH):
    raise FileNotFoundError(f"DSM not found: {DSM_PATH}")

#-------------------- OPEN DSM FILE AND READ DATA --------------------
with rasterio.open(DSM_PATH) as src:
    dsm_band = src.read(1)
    nodata = src.nodatavals[0]
    dsm_crs = src.crs
    dsm_transform = src.transform
    bounds = src.bounds

# ---------- CONFIG ----------
meta_path  = "extracted_images2/metadata_with_euler.csv"

# ---------- LOAD METADATA ----------
meta_data = pd.read_csv(meta_path)

required_cols = {
    "filename", "gps_latitude", "gps_longitude", "gps_altitude",
    "roll", "pitch", "yaw",
    "imu_orientation_x", "imu_orientation_y", "imu_orientation_z", "imu_orientation_w"
}
missing = required_cols - set(meta_data.columns)
if missing:
    raise KeyError(f"Missing metadata columns: {missing}")

# ---------- PREPARE DYNAMIC SIZES ----------------
# Collect all positions first to compute trajectory extent
all_easting = []
all_northing = []
transformer = Transformer.from_crs("EPSG:4326", dsm_crs, always_xy=True)

for uav_imag_num in range(1, 1327):


    row = meta_data.iloc[uav_imag_num]
    uav_lat_gps  = float(row["gps_latitude"])
    uav_lon_gps  = float(row["gps_longitude"])
    uav_e_dsm, uav_n_dsm = transformer.transform(uav_lon_gps, uav_lat_gps)
    all_easting.append(uav_e_dsm)
    all_northing.append(uav_n_dsm)

all_easting = np.array(all_easting)
all_northing = np.array(all_northing)

# Compute trajectory size for scaling points and arrows
traj_size = np.sqrt((all_easting.max() - all_easting.min())**2 +
                    (all_northing.max() - all_northing.min())**2)
arrow_length = traj_size * 0.01       # 1% of trajectory length
arrow_head   = arrow_length * 0.5     # proportional head size
marker_size  = traj_size * 0.002      # bigger points



# ---------- PLOT TRAJECTORY ----------
plt.figure(figsize=(12,12))


# Plot the trajectory as a connected line
plt.plot(all_easting, all_northing, color='green', linewidth=1, label="Trajectory")
# Mark start and end points
plt.scatter(all_easting[0], all_northing[0], color='lime', s=200, marker='o', label='Start')  # start point
plt.scatter(all_easting[-1], all_northing[-1], color='magenta', s=200, marker='X', label='End')  # end point

skip = 10
for uav_imag_num in range(1, 1327):
    if uav_imag_num % skip != 0:
        continue
    
    if not (0 <= uav_imag_num < len(meta_data)):
        raise IndexError(f"uav_imag_num {uav_imag_num} out of range")

    row = meta_data.iloc[uav_imag_num]

    # ---------- EXTRACT METADATA ----------
    uav_lat_gps  = float(row["gps_latitude"])
    uav_lon_gps  = float(row["gps_longitude"])
    uav_alt_gps  = float(row["gps_altitude"])
    uav_yaw      = float(row["yaw"])
    qx = float(row["imu_orientation_x"])
    qy = float(row["imu_orientation_y"])
    qz = float(row["imu_orientation_z"])
    qw = float(row["imu_orientation_w"])

    #-------------------- TRANSFORM UAV GPS COORDINATES TO DSM CRS --------------------
    uav_e_dsm, uav_n_dsm = transformer.transform(uav_lon_gps, uav_lat_gps)

    # Plot UAV position
    plt.scatter(uav_e_dsm, uav_n_dsm, s=marker_size*100, marker='.', color="r")
    plt.plot(uav_e_dsm, uav_n_dsm, 'g')

    # Plot heading arrow
    dx = arrow_length * np.cos(uav_yaw)
    dy = arrow_length * np.sin(uav_yaw)
    plt.arrow(uav_e_dsm, uav_n_dsm, dx, dy, color='blue', head_width=arrow_head, head_length=arrow_head)



# ---------- FINALIZE PLOT ----------
plt.xlabel("Easting (Map CRS)")
plt.ylabel("Northing (Map CRS)")
plt.axis('equal')
plt.grid(True)
plt.legend()
plt.savefig('trajectory.png')
plt.show()
```

```python
import os
import cv2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pyproj import Transformer
import rasterio

#-------------------- PATH --------------------
DSM_PATH = "../shared/datasets/garrulus-arnsberg-forest/data_collection/2025-11-19_field-D-E/post-processing-results/2025-11-19_field-D_flight-10_rgb-w-gcp/rgb_wide_images-only_dtm.tif"

#-------------------- CHECK DSM FILE --------------------
if not os.path.exists(DSM_PATH):
    raise FileNotFoundError(f"DSM not found: {DSM_PATH}")

#-------------------- OPEN DSM FILE AND READ DATA --------------------
with rasterio.open(DSM_PATH) as src:
    dsm_band = src.read(1)
    nodata = src.nodatavals[0]
    dsm_crs = src.crs
    dsm_transform = src.transform
    bounds = src.bounds

# ---------- CONFIG ----------
meta_path  = "extracted_images2/metadata_with_euler.csv"

# ---------- LOAD METADATA ----------
meta_data = pd.read_csv(meta_path)

required_cols = {
    "filename", "gps_latitude", "gps_longitude", "gps_altitude",
    "roll", "pitch", "yaw",
    "imu_orientation_x", "imu_orientation_y", "imu_orientation_z", "imu_orientation_w"
}
missing = required_cols - set(meta_data.columns)
if missing:
    raise KeyError(f"Missing metadata columns: {missing}")

# ---------- PREPARE DYNAMIC SIZES ----------------
all_easting = []
all_northing = []
transformer = Transformer.from_crs("EPSG:4326", dsm_crs, always_xy=True)

for uav_imag_num in range(1, 1327):
    row = meta_data.iloc[uav_imag_num]
    uav_lat_gps  = float(row["gps_latitude"])
    uav_lon_gps  = float(row["gps_longitude"])
    uav_e_dsm, uav_n_dsm = transformer.transform(uav_lon_gps, uav_lat_gps)
    all_easting.append(uav_e_dsm)
    all_northing.append(uav_n_dsm)

all_easting = np.array(all_easting)
all_northing = np.array(all_northing)

traj_size = np.sqrt((all_easting.max() - all_easting.min())**2 +
                    (all_northing.max() - all_northing.min())**2)

arrow_length = traj_size * 0.02      # BIGGER arrows (was 0.01)
arrow_head   = arrow_length * 0.8
marker_size  = traj_size * 0.004     # bigger points

# ---------- PLOT TRAJECTORY ----------
plt.figure(figsize=(14,14))

# Trajectory line (thicker)
plt.plot(all_easting, all_northing, color='green', linewidth=2.5, label="Trajectory")

# Start & End (much bigger + outlined)
plt.scatter(all_easting[0], all_northing[0],
            color='lime', s=300, edgecolors='black', linewidth=2, label='Start')

plt.scatter(all_easting[-1], all_northing[-1],
            color='magenta', s=300, marker='X', edgecolors='black', linewidth=2, label='End')

# LABELS (large + readable)
plt.text(all_easting[0], all_northing[0], " START",
         fontsize=14, weight='bold',
         bbox=dict(facecolor='white', alpha=0.8, edgecolor='black'))

plt.text(all_easting[-1], all_northing[-1], " END",
         fontsize=14, weight='bold',
         bbox=dict(facecolor='white', alpha=0.8, edgecolor='black'))

skip = 10
for uav_imag_num in range(1, 1327):
    if uav_imag_num % skip != 0:
        continue

    row = meta_data.iloc[uav_imag_num]

    uav_lat_gps  = float(row["gps_latitude"])
    uav_lon_gps  = float(row["gps_longitude"])
    uav_yaw      = float(row["yaw"])

    uav_e_dsm, uav_n_dsm = transformer.transform(uav_lon_gps, uav_lat_gps)

    # UAV points (bigger)
    plt.scatter(uav_e_dsm, uav_n_dsm,
                s=marker_size*200,
                color="red",
                edgecolors='black',
                linewidth=0.5)

    # STRONG ARROWS using quiver
    dx = arrow_length * np.cos(uav_yaw)
    dy = arrow_length * np.sin(uav_yaw)

    plt.quiver(uav_e_dsm, uav_n_dsm, dx, dy,
               angles='xy', scale_units='xy', scale=1,
               color='blue',
               width=0.004,           # thicker shaft
               headwidth=6,
               headlength=8)

# ---------- FINALIZE ----------
plt.xlabel("Easting (Map CRS)", fontsize=14)
plt.ylabel("Northing (Map CRS)", fontsize=14)
plt.title("UAV Trajectory with Heading", fontsize=16, weight='bold')

plt.axis('equal')
plt.grid(True, linewidth=0.5)

plt.legend(fontsize=12)
plt.savefig('trajectory.png', dpi=300)
plt.show()
```

```python
import os
import cv2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pyproj import Transformer
import rasterio

#-------------------- PATH --------------------
DSM_PATH = "../shared/datasets/garrulus-arnsberg-forest/data_collection/2025-11-19_field-D-E/post-processing-results/2025-11-19_field-D_flight-10_rgb-w-gcp/rgb_wide_images-only_dtm.tif"

if not os.path.exists(DSM_PATH):
    raise FileNotFoundError(f"DSM not found: {DSM_PATH}")

with rasterio.open(DSM_PATH) as src:
    dsm_band = src.read(1)
    dsm_crs = src.crs

# ---------- CONFIG ----------
meta_path  = "extracted_images2/metadata_with_euler.csv"
meta_data = pd.read_csv(meta_path)

transformer = Transformer.from_crs("EPSG:4326", dsm_crs, always_xy=True)

# ---------- TRAJECTORY ----------
all_easting, all_northing = [], []

for i in range(1, 1327):
    row = meta_data.iloc[i]
    e, n = transformer.transform(row["gps_longitude"], row["gps_latitude"])
    all_easting.append(e)
    all_northing.append(n)

all_easting = np.array(all_easting)
all_northing = np.array(all_northing)

traj_size = np.sqrt((all_easting.max() - all_easting.min())**2 +
                    (all_northing.max() - all_northing.min())**2)

arrow_length = traj_size * 0.03

# Offset for labels (so they don't cover markers)
offset = traj_size * 0.01

# ---------- PLOT ----------
plt.figure(figsize=(14,14))

# Trajectory
plt.plot(all_easting, all_northing,
         color='green', linewidth=3, label="Trajectory", zorder=1)

# Start / End markers (HIGH zorder so they stay visible)
plt.scatter(all_easting[0], all_northing[0],
            color='lime', s=350, edgecolors='black', linewidth=2,
            label='Start', zorder=5)

plt.scatter(all_easting[-1], all_northing[-1],
            color='magenta', s=350, marker='X', edgecolors='black', linewidth=2,
            label='End', zorder=5)

# ---------- FIXED LABEL POSITIONS ----------
plt.text(all_easting[0] + offset, all_northing[0] + offset,
         "START",
         fontsize=16, weight='bold',
         bbox=dict(facecolor='white', alpha=0.9, edgecolor='black'),
         zorder=4)

plt.text(all_easting[-1] + offset, all_northing[-1] + offset,
         "END",
         fontsize=16, weight='bold',
         bbox=dict(facecolor='white', alpha=0.9, edgecolor='black'),
         zorder=4)

# ---------- ARROWS ----------
skip = 40

for i in range(1, 1327):
    if i % skip != 0:
        continue

    row = meta_data.iloc[i]

    e, n = transformer.transform(row["gps_longitude"], row["gps_latitude"])
    yaw = float(row["yaw"])

    dx = arrow_length * np.cos(yaw)
    dy = arrow_length * np.sin(yaw)

    plt.quiver(e, n, dx, dy,
               angles='xy', scale_units='xy', scale=1,
               color='blue',
               width=0.004,
               headwidth=14,
               headlength=18,
               headaxislength=14,
               zorder=2)

# ---------- FINAL ----------
plt.xlabel("Easting", fontsize=20)
plt.ylabel("Northing", fontsize=20)
plt.title("UAV Trajectory", fontsize=20)

plt.axis('equal')
plt.grid(True)

# ---------- SPACED LEGEND ----------
plt.legend(
    loc='upper left',
    fontsize=20,
    markerscale=1.5,
    frameon=True,
    facecolor='white',
    edgecolor='black',
    labelspacing=1.4,
    handletextpad=1.0,
    borderpad=1.2
)

plt.tight_layout()
plt.savefig('trajectory.png', dpi=300)
plt.show()
```
